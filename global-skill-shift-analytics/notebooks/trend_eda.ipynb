```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Load data

```python
df = pd.read_csv('../data/google_trends.csv')
```

Preview

```python
print(df.head())
```

```python
print(df.info())
```

Convert 'date' column to datetime if not already

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
print(df.head())
```

Plotting trend of each skill

```python
skills = df.columns.drop('date')
plt.figure(figsize=(15, 8))

for skill in skills:
    plt.plot(df['date'], df[skill], label=skill)

plt.title('Google Search Trends for Tech Skills')
plt.xlabel('Date')
plt.ylabel('Search Interest')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.grid(True)
plt.show()
```
